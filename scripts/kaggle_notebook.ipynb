# BISINDO WL15 - training session

Settings (right panel): **Accelerator = GPU T4 x2** (only GPU 0 is used, no DDP), **Internet = on**, **Persistence = off**.
Inputs: the private dataset `bisindo-wl15`; from the second session on, also the **output of the previous version of this notebook** (Add Input > Your Work > this notebook).

Run with **Save Version > Save & Run All (Commit)**. Training stops by itself before `MAX_HOURS` (default 10.8 h); checkpoints are saved every epoch and the next session resumes them.

```python
import glob, subprocess, sys
from pathlib import Path

PLAN = "main"
# Session length limit in hours, counted from the start of the session script. No epoch is started that would
# end after it; the report and the output save take a few more minutes. Lower it when less than ~11 h of the
# weekly Kaggle GPU quota is left (e.g. 6.5 when 7 h are left), so the session is not cut off by the quota.
MAX_HOURS = 10.8

# the dataset copy of the session script (not the copies inside an attached earlier output)
cands = [p for p in glob.glob("/kaggle/input/**/scripts/kaggle_session.py", recursive=True)
         if (Path(p).parents[1] / "manifest_sha256.txt").exists()]
assert len(cands) == 1, cands
print("session script:", cands[0])
# The child's output is piped and printed here, line by line. A child writing straight to the kernel's
# stdout appears several times in the saved Kaggle log (the raw process stream and the cell output).
proc = subprocess.Popen([sys.executable, "-u", cands[0], "--plan", PLAN, "--max-hours", str(MAX_HOURS)],
                        stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, errors="replace", bufsize=1)
for line in proc.stdout:
    print(line, end="", flush=True)
print("exit code:", proc.wait())
```
